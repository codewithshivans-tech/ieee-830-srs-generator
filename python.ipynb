```python
import datetime

def generate_srs(
    project_name="AI SRS Document Generator",
    task_id="AI-PR-004",
    student_code="DAS013618",
    style="IEEE 830 Standard",
    description="An automated tool that generates structured Software Requirements Specifications (SRS)."
):
    styles_config = {
        "IEEE 830 Standard": {
            "tone": "Formal & Technical",
            "sections": ["1. Introduction", "2. Overall Description", "3. Specific Requirements", "4. Quality & Compliance"]
        },
        "Agile / concise": {
            "tone": "Action-Oriented & Clear",
            "sections": ["1. Overview & Goal", "2. Core Features", "3. User & System Requirements", "4. Acceptance Criteria"]
        },
        "Detailed Enterprise": {
            "tone": "Comprehensive & Audit-Ready",
            "sections": ["1. Executive Summary", "2. System Context", "3. Functional & Non-Functional Specs", "4. Risk, Security & Quality"]
        }
    }

    selected = styles_config.get(style, styles_config["IEEE 830 Standard"])
    now = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")

    # Simple completeness / quality scoring logic (out of 100)
    score = 60
    if len(project_name.strip()) > 3:
        score += 10
    if len(description.strip()) >= 30:
        score += 15
    if style in styles_config:
        score += 15
    score = min(score, 100)

    srs_content = f"""# Software Requirements Specification (SRS)
**Project Name:** {project_name}
**Task ID:** {task_id}
**Student Code:** {student_code}
**Format Style:** {style} ({selected["tone"]})
**Generated On:** {now}
**SRS Quality Score:** {score}/100

---

## {selected["sections"][0]}
### 1.1 Purpose
This document specifies the functional and non-functional requirements for **{project_name}** following the **{style}** approach.

### 1.2 Scope
{description}

### 1.3 Definitions & Standards
- **IEEE 830:** Standard recommended practice for Software Requirements Specifications.
- **SRS:** A complete description of behavior of a system to be developed.

---

## {selected["sections"][1]}
### 2.1 Product Perspective
The system acts as a standalone requirement-drafting utility that turns project inputs into a structured Markdown SRS document.

### 2.2 User Classes & Characteristics
- **Students & Developers:** Drafting project specifications quickly.
- **Project Reviewers:** Evaluating requirement clarity, structure, and completeness.

### 2.3 Operating Environment
- Python 3.x runtime (Google Colab or local environment).
- Markdown-compatible viewer or GitHub repository.

---

## {selected["sections"][2]}
### 3.1 Functional Requirements
- **FR-01:** Accept project title, description, task metadata, and SRS style selection.
- **FR-02:** Generate structured sections aligned with IEEE 830 guidelines.
- **FR-03:** Support multiple SRS formatting styles (`IEEE 830 Standard`, `Agile / concise`, `Detailed Enterprise`).
- **FR-04:** Export the generated specification to `SRS_Document.md`.

### 3.2 Non-Functional Requirements
- **NFR-01 Usability:** Clear headings and readable Markdown formatting.
- **NFR-02 Performance:** Generates output in under 2 seconds.
- **NFR-03 Maintainability:** Modular Python function design for easy extension.

---

## {selected["sections"][3]}
### 4.1 Quality Scoring Breakdown
- **Completeness & Clarity Score:** **{score}/100**
- **Structure Check:** IEEE 830 core sections included (Introduction, Overall Description, Specific Requirements, Quality/Verification).

### 4.2 Verification Criteria
- Verify `SRS_Document.md` is created without syntax or runtime errors.
- Confirm all required sections and requirement IDs (`FR-01`, `NFR-01`) are present.
"""

    with open("SRS_Document.md", "w", encoding="utf-8") as f:
        f.write(srs_content)

    print(f"SRS generated successfully! Quality Score: {score}/100")
    print("Saved file: SRS_Document.md")
    return srs_content


if __name__ == "__main__":
    print("--- Available Styles ---")
    print("1. IEEE 830 Standard")
    print("2. Agile / concise")
    print("3. Detailed Enterprise")

    proj = input("Enter Project Name [AI SRS Document Generator]: ").strip() or "AI SRS Document Generator"
    desc = input("Enter Project Description: ").strip()
    if not desc:
        desc = "Automated IEEE 830 compliant Software Requirements Specification generator built with Python."

    choice = input("Choose style (1/2/3) [1]: ").strip()
    style_map = {
        "1": "IEEE 830 Standard",
        "2": "Agile / concise",
        "3": "Detailed Enterprise"
    }
    chosen_style = style_map.get(choice, "IEEE 830 Standard")

    output = generate_srs(
        project_name=proj,
        task_id="AI-PR-004",
        student_code="DAS013618",
        style=chosen_style,
        description=desc
    )
    print("\n--- Preview (First 600 chars) ---\n")
    print(output[:600])
```
